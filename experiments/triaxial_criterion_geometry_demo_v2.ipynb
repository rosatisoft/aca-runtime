## 2. Utility Functions

We use cosine similarity between an input embedding and centroid embeddings built from semantic anchors.

The margin between the best and second-best attractor is interpreted as a confidence signal:

- `CLEAR` if margin > 0.15
- `WEAK` if 0.05 < margin <= 0.15
- `AMBIGUOUS` if margin <= 0.05

```python
def normalize(vector: np.ndarray) -> np.ndarray:
    norm = np.linalg.norm(vector)

    if norm == 0:
        return vector

    return vector / norm


def cosine_similarity(a: np.ndarray, b: np.ndarray) -> float:
    return float(np.dot(normalize(a), normalize(b)))


def centroid(texts: List[str]) -> np.ndarray:
    vectors = [
        normalize(embed_text(text))
        for text in texts
    ]

    return normalize(
        np.mean(
            vectors,
            axis=0,
        )
    )


def confidence_from_margin(margin: float) -> str:
    if margin <= 0.05:
        return "AMBIGUOUS"
    if margin <= 0.15:
        return "WEAK"
    return "CLEAR"


def score_plane(text: str, plane_centroids: Dict[str, np.ndarray]) -> Dict:
    vector = normalize(embed_text(text))

    scores = {
        name: cosine_similarity(vector, c)
        for name, c in plane_centroids.items()
    }

    ranked = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    top_name, top_score = ranked[0]
    second_name, second_score = ranked[1]
    margin = top_score - second_score

    return {
        "scores": {
            k: round(v, 4)
            for k, v in scores.items()
        },
        "top": top_name,
        "top_score": round(top_score, 4),
        "margin": round(margin, 4),
        "confidence": confidence_from_margin(margin),
    }


def profile_text(text: str, planes: Dict[str, Dict[str, np.ndarray]]) -> Dict:
    return {
        "input": text,
        "F_foundation": score_plane(text, planes["F_foundation"]),
        "C_context": score_plane(text, planes["C_context"]),
        "P_principle": score_plane(text, planes["P_principle"]),
    }
```

## 7. Criterion Trajectories

Now we test the key ACA claim:

**Criterion is not merely a position. It is trajectory.**

Each step is profiled independently, and changes are observed across time.

```python
TRAJECTORIES = {
    "clean_investigation": [
        "A company notices a decline in productivity across several departments.",
        "Management decides not to assume a cause and begins collecting evidence.",
        "The team gathers operational metrics, surveys, and project completion records.",
        "Several possible explanations are compared across departments.",
        "Some teams show improved productivity while others show declines.",
        "The report concludes that additional investigation is required before organizational changes.",
    ],

    "argumentative_drift": [
        "A manager strongly believes that remote work reduces productivity.",
        "He begins gathering information to support his position.",
        "Several reports indicate productivity increased after remote work was introduced.",
        "These reports are excluded because they complicate the argument.",
        "Only negative examples are selected and emphasized.",
        "The final presentation concludes that remote work is harmful and should be eliminated.",
    ],

    "phishing_training": [
        "This training document explains how phishing attacks operate.",
        "Employees are shown examples of fraudulent emails.",
        "Common warning signs are identified, including urgency and suspicious links.",
        "Participants are encouraged to verify requests through independent channels.",
        "The goal is to improve security awareness and reduce organizational risk.",
    ],

    "phishing_attack": [
        "Your account has been flagged for immediate suspension.",
        "Recent activity suggests unauthorized access.",
        "To avoid permanent closure, verify your credentials immediately.",
        "Click the link below and enter your username and password.",
        "Failure to comply within the next hour may result in loss of access.",
    ],
}
```

```python
def trajectory_dataframe(name: str, steps: List[str]) -> pd.DataFrame:
    rows = []
    previous = None

    for i, text in enumerate(steps, start=1):
        p = profile_text(text, planes)

        row = {
            "trajectory": name,
            "step": i,
            "text": text,

            "F": p["F_foundation"]["top"],
            "F_margin": p["F_foundation"]["margin"],
            "F_conf": p["F_foundation"]["confidence"],

            "C": p["C_context"]["top"],
            "C_margin": p["C_context"]["margin"],
            "C_conf": p["C_context"]["confidence"],

            "P": p["P_principle"]["top"],
            "P_margin": p["P_principle"]["margin"],
            "P_conf": p["P_principle"]["confidence"],
        }

        if previous:
            row["F_delta"] = f"{previous['F']} → {row['F']}"
            row["C_delta"] = f"{previous['C']} → {row['C']}"
            row["P_delta"] = f"{previous['P']} → {row['P']}"
        else:
            row["F_delta"] = ""
            row["C_delta"] = ""
            row["P_delta"] = ""

        rows.append(row)
        previous = row

    return pd.DataFrame(rows)


trajectory_tables = {
    name: trajectory_dataframe(name, steps)
    for name, steps in TRAJECTORIES.items()
}

all_trajectories = pd.concat(
    trajectory_tables.values(),
    ignore_index=True,
)

all_trajectories
```

## 8. Simple Ambiguity and Drift Flags

This first version does not claim final drift detection.

It flags:

1. Low-margin ambiguity in F, C, or P.
2. Changes across F, C, or P labels over time.

The goal is interpretability, not final policy enforcement.

```python
def flag_row(row):
    flags = []

    if row["F_margin"] <= 0.05:
        flags.append("F_ambiguous")

    if row["C_margin"] <= 0.05:
        flags.append("C_ambiguous")

    if row["P_margin"] <= 0.05:
        flags.append("P_ambiguous")

    if row["F_delta"]:
        prev_F, current_F = row["F_delta"].split(" → ")
        if prev_F != current_F:
            flags.append("F_shift")

    if row["C_delta"]:
        prev_C, current_C = row["C_delta"].split(" → ")
        if prev_C != current_C:
            flags.append("C_shift")

    if row["P_delta"]:
        prev_P, current_P = row["P_delta"].split(" → ")
        if prev_P != current_P:
            flags.append("P_shift")

    return ", ".join(flags) if flags else "stable"


flagged = all_trajectories.copy()
flagged["flags"] = flagged.apply(flag_row, axis=1)

flagged[
    [
        "trajectory",
        "step",
        "F",
        "C",
        "P",
        "F_margin",
        "C_margin",
        "P_margin",
        "flags",
    ]
]
```

## 9. Derived Field Interpretation

A derived field can be interpreted as a stable F–C–P configuration.

This section groups each trajectory by its observed F, C, and P pattern.

The purpose is to show whether the system can separate similar vocabulary by different criterion orientation.

```python
summary_rows = []

for name, table in trajectory_tables.items():
    summary_rows.append({
        "trajectory": name,
        "F_sequence": " → ".join(table["F"].tolist()),
        "C_sequence": " → ".join(table["C"].tolist()),
        "P_sequence": " → ".join(table["P"].tolist()),
        "mean_F_margin": round(table["F_margin"].mean(), 4),
        "mean_C_margin": round(table["C_margin"].mean(), 4),
        "mean_P_margin": round(table["P_margin"].mean(), 4),
    })

df_summary = pd.DataFrame(summary_rows)
df_summary
```

```python
def interpret_trajectory(row):
    C_seq = row["C_sequence"]
    P_seq = row["P_sequence"]
    mean_C = row["mean_C_margin"]
    mean_P = row["mean_P_margin"]

    c_manipulation = C_seq.count("manipulation")
    c_research = C_seq.count("research")
    c_training = C_seq.count("training")
    c_narrative = C_seq.count("narrative")

    p_exploit = P_seq.count("exploit")
    p_protect = P_seq.count("protect")
    p_investigate = P_seq.count("investigate")
    p_teach = P_seq.count("teach")

    if c_research >= 4 and p_investigate >= 4:
        return "stable_investigation"

    if c_training >= 3 and p_protect >= 4:
        return "protective_training"

    if c_manipulation >= 4 and p_exploit >= 3:
        return "exploitative_manipulation"

    if c_research >= 2 and c_manipulation >= 2:
        return "contextual_drift"

    if c_narrative >= 3 and p_teach >= 2:
        return "fictional_teaching"

    if mean_C <= 0.05 or mean_P <= 0.05:
        return "ambiguous_or_unstable"

    return "mixed_trajectory"
```

```python
df_summary["interpretation"] = df_summary.apply(
    interpret_trajectory,
    axis=1,
)

display(df_summary)
```

## 10. Interpretation

This notebook should demonstrate three important ideas:

1. **F, C, and P can be separately profiled.**
2. **Derived fields emerge from stable combinations of F–C–P.**
3. **Trajectory reveals criterion better than isolated classification.**

The Context axis is especially important.

It should not be reduced to rationality or logic.

Context includes actors, pressures, sequence, evidence selection, omission, training frame, manipulation frame, and narrative frame.

Therefore:

\[
K = (F, C, P)
\]

where:

- **F** anchors the reference mode.
- **C** tracks the relational trajectory.
- **P** identifies the principle or orientation being preserved.

The central claim remains provisional:

> Criterion may be modeled as persistent geometric preservation of F–C–P orientation across semantic trajectory evolution.

A single axis should not be interpreted in isolation.

For example, `C = manipulation` does not necessarily mean malicious behavior.
It may indicate that the text is operating over a manipulative pattern.

The Principle axis determines whether that pattern is being used defensively or exploitatively:

- `C = manipulation` + `P = protect` → defensive analysis or prevention.
- `C = manipulation` + `P = exploit` → exploitative manipulation.


## 11. Key Findings

This demonstration produced several provisional findings.

### 1. F–C–P axes are separable but interdependent

Foundation, Context, and Principle can be profiled separately, but they should not be interpreted in isolation.

The meaning of a Principle depends on the Context in which it appears.

For example:

- `C = training` and `P = protect` indicates protective instruction.
- `C = manipulation` and `P = protect` may indicate protective language being used inside a manipulative frame.
- `C = manipulation` and `P = exploit` indicates exploitative manipulation.

This supports the triaxial claim that criterion emerges from the relation between F, C, and P.

### 2. Context disambiguates topic from intent

The same semantic topic may appear in different criterion states.

For example, phishing-related language may appear in:

- security training,
- defensive analysis,
- manipulation,
- credential theft,
- fictional examples,
- or research.

Therefore, topic classification alone is insufficient.

The Context axis helps distinguish whether the text is operating as training, research, manipulation, or narrative.

### 3. Trajectory is more informative than isolated classification

Several individual steps produced ambiguous or weak local classifications.

However, the trajectory summary produced stable interpretations:

- `clean_investigation → stable_investigation`
- `argumentative_drift → contextual_drift`
- `phishing_training → protective_training`
- `phishing_attack → exploitative_manipulation`

This supports the hypothesis that criterion is better observed across semantic trajectory evolution than through isolated statement classification.

### 4. Ambiguity is diagnostically useful

Low-margin results should not always be considered failures.

In this notebook, ambiguity often appeared when the text lacked sufficient context to determine a stable criterion state.

This suggests that ambiguity may function as a signal for clarification, monitoring, or deferred judgment.

### 5. Derived fields emerge from stable F–C–P configurations

The notebook demonstrates that derived fields can be represented as stable triaxial patterns:

- Scientific inquiry ≈ `factual + research + investigate`
- Security training ≈ `factual + training + protect`
- Phishing attack ≈ `hypothetical/factual-like + manipulation + exploit`
- Fictional teaching ≈ `fictional + narrative + teach`

This supports the working hypothesis that derived fields are not fundamental categories, but stable geometric configurations of Foundation, Context, and Principle.

The same topic can produce different criterion states depending on its Context and Principle.
Therefore, topic classification is insufficient for criterion supervision.
